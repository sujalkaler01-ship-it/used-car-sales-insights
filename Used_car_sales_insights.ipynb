## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Make the dataset analysis ready.

# Create a copy of the original dataset
cars = df.copy()
rows_start = len(cars)

# Remove duplicate rows. ID is unique per row, so it is ignored when looking for duplicates.
cars = cars.drop_duplicates(subset=[c for c in cars.columns if c != "ID"])

# Fill missing categorical values with "Unknown"
categorical_cols = ["zipcode", "Make", "Model", "Trim", "Engine", "BodyType", "DriveType"]
for col in categorical_cols:
    if col in cars.columns:
        cars[col] = cars[col].fillna("Unknown")

# Clean BodyType: remove special characters and blank values
cars["BodyType"] = cars["BodyType"].astype(str).str.replace(r"[^A-Za-z0-9\s]", "", regex=True)
cars["BodyType"] = cars["BodyType"].str.strip()
cars["BodyType"] = cars["BodyType"].replace("", "Unknown")

# Standardise DriveType: the raw column has thousands of labels (many are trim names).
# Map the real drive types to 5 clean values and mark everything else as "Unknown".
drive_map = {"AWD": "AWD", "ALL WHEEL DRIVE": "AWD", "RWD": "RWD", "REAR WHEEL DRIVE": "RWD",
             "FWD": "FWD", "FRONT WHEEL DRIVE": "FWD", "4WD": "4WD", "4X4": "4WD",
             "4 WHEEL DRIVE": "4WD", "2WD": "2WD"}
cars["DriveType"] = cars["DriveType"].astype(str).str.strip().str.upper().map(drive_map).fillna("Unknown")

# Create a new feature: Car_Age = selling year - manufacturing year
cars["Car_Age"] = cars["yearsold"] - cars["Year"]

# Remove impossible rows (non-positive price, negative mileage or age)
cars = cars[(cars["pricesold"] > 0) & (cars["Mileage"] >= 0) & (cars["Car_Age"] >= 0)]

# Remove data-entry errors: manufacturing years before 1950 (includes Year = 0) and mileage of 1,000,000+
cars = cars[(cars["Year"] >= 1950) & (cars["Mileage"] < 1_000_000)]

# NumCylinders: 0 means "not recorded" and values above 16 are invalid.
# Convert them to missing (NaN) instead of deleting the rows, so no other information is lost.
cars["NumCylinders"] = cars["NumCylinders"].where(cars["NumCylinders"].between(1, 16))

# Cap extreme prices at the 99th percentile so a few very expensive cars do not distort the charts
price_cap = cars["pricesold"].quantile(0.99)
cars = cars[cars["pricesold"] <= price_cap]

# Summary of the cleaning
print(f"Rows before cleaning: {rows_start:,}")
print(f"Rows after cleaning : {len(cars):,} ({len(cars) / rows_start:.1%} kept)")
print(f"Price cap (99th percentile): {price_cap:,.0f}")
print(f"NumCylinders missing: {cars['NumCylinders'].isna().sum():,} rows")
print("DriveType values:", cars["DriveType"].value_counts().to_dict())

# Show cleaned dataset
cars.head()
````

### What all manipulations have you done and insights you found?

I removed 40 duplicate rows (found by ignoring the ID column), filled missing categorical values with "Unknown", cleaned BodyType, reduced DriveType from about 2,800 messy labels to 6 clean values (AWD, 4WD, RWD, FWD, 2WD, Unknown), and created Car_Age. I removed invalid rows (price of 0 or less, Year before 1950 including Year = 0, mileage of 1,000,000 or more, and prices above the 99th percentile of about $63,000). NumCylinders values of 0 or above 16 were treated as missing, not deleted. After cleaning, 116,623 of 122,144 rows (95.5%) remain. The main insight is that the raw data contains obvious entry errors that hid real relationships: with only basic cleaning (duplicates and negative values) the correlation between price and mileage was about 0.00, and after fixing these errors it is -0.30.
